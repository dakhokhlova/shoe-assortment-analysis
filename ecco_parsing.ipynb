```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}
```

```python
def get_soup(url):
    
    response = requests.get(url, headers=headers)
    
    if response.status_code != 200:
        return None
    
    soup = BeautifulSoup(response.text, 'html.parser')
    return soup
```

```python
def parse_page(soup):
    cards = soup.select('div.product-card')
    page_data = []
    for card in cards:
        name = card.get('data-name')
        if not name:
            continue
        
        item = {
            'name': name,
            'price': int(card.get('data-price') or 0),
            'price_old': int(card.get('data-price-old') or 0),
            'type': card.get('data-type'),
            'gender': card.get('data-category'),
            'sizes': [size.get_text(strip=True) for size in card.select('.sizes .size')],
            'color': card.get('data-variant'),
            'article': card.get('data-art-id'),
            'collection': card.get('data-collection')
        }
        page_data.append(item)
    return page_data

all_data = []

soup = get_soup('https://www.ecco.ru/women/shoes/all/')
if soup:
    all_data.extend(parse_page(soup))

page = 2
while page < 9:
    url = f'https://www.ecco.ru/women/shoes/all/?p={page}'
    soup = get_soup(url)
    if soup is None:
        break
    all_data.extend(parse_page(soup))
    page += 1
```

```python
df_women = pd.DataFrame(all_data)
df_women.head()
```

```python
all_data = []

soup = get_soup('https://www.ecco.ru/men/shoes/all/')
if soup:
    all_data.extend(parse_page(soup))

page = 2
while page < 8:
    url = f'https://www.ecco.ru/men/shoes/all/?p={page}'
    soup = get_soup(url)
    if soup is None:
        break
    all_data.extend(parse_page(soup))
    page += 1
```

```python
df_men = pd.DataFrame(all_data)
df_men.head()
```

```python
all_data = []

soup = get_soup('https://www.ecco.ru/kids/')
if soup:
    all_data.extend(parse_page(soup))

page = 2
while page < 5:
    url = f'https://www.ecco.ru/kids/?p={page}'
    soup = get_soup(url)
    if soup is None:
        break
    all_data.extend(parse_page(soup))
    page += 1
```

```python
df_kids = pd.DataFrame(all_data)
df_kids.head()
```

```python
df_all = pd.concat([df_women, df_men, df_kids], ignore_index=True)
```
